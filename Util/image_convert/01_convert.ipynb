이미지 파일 위치 이동코드

```python
import os
import random
import shutil
from tqdm import tqdm

def random_sample_images(src_folder, dst_folder, sample_size, extensions=(".jpg", ".png", ".jpeg")):
    # 원본 폴더에서 이미지 목록 가져오기
    images = [f for f in os.listdir(src_folder) if f.lower().endswith(extensions)]
    
    if sample_size > len(images):
        raise ValueError(f"요청한 개수({sample_size})가 이미지 개수({len(images)})보다 많습니다.")
    
    # 무작위로 이미지 선택
    sampled = random.sample(images, sample_size)
    
    # 결과 저장 폴더 생성
    os.makedirs(dst_folder, exist_ok=True)
    
    # 선택된 이미지 복사
    for img in tqdm(sampled):
        src_path = os.path.join(src_folder, img)
        dst_path = os.path.join(dst_folder, img)
        shutil.move(src_path, dst_path)
    
    print(f"{sample_size}개의 이미지를 {dst_folder}에 저장했습니다.")

# 사용 예시
src = "/home/addinedu/dev_ws/deep/image_setting/original_data/Phone"      # 원본 이미지 폴더
dst = "/home/addinedu/dev_ws/deep/image_setting/extract_data/Phone_50_3"  # 추출한 이미지 저장 폴더

sample_size = 50 # 이동할 이미지 수량
random_sample_images(src, dst, sample_size)
```

아이폰 이미지 HEIC -> png 변환

```python
import os
from PIL import Image
import pillow_heif
from tqdm import tqdm

def convert_heic_to_png(src_folder, dst_folder):
    # 출력 폴더 생성
    os.makedirs(dst_folder, exist_ok=True)
    
    # 폴더 내 모든 파일 탐색
    for filename in tqdm(os.listdir(src_folder)):
        if filename.lower().endswith(".heic"):
            heic_path = os.path.join(src_folder, filename)
            png_filename = os.path.splitext(filename)[0] + ".png"
            png_path = os.path.join(dst_folder, png_filename)
            
            # HEIC → PIL Image 변환
            heif_file = pillow_heif.read_heif(heic_path)
            image = Image.frombytes(
                heif_file.mode,
                heif_file.size,
                heif_file.data,
                "raw"
            )
            
            # PNG 저장
            image.save(png_path, "PNG")
            print(f"변환 완료: {png_path}")

# 사용 예시
src = "/home/addinedu/dev_ws/deep/image_setting/original_data/earphone"     # HEIC 이미지 폴더
dst = "/home/addinedu/dev_ws/deep/image_setting/extract_data/Earphone"      # 변환된 PNG 저장 폴더
convert_heic_to_png(src, dst)
```
