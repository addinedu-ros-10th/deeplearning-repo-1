```python
import cv2
from ultralytics import YOLO

model = YOLO("./runs/detect/json_ai_848_5items_sdj_v1.0/weights/best.pt")

cap = cv2.VideoCapture(0)

while cap.isOpened():

    ret, frame = cap.read()

    if not ret:
        break

    result = model(frame, conf = 0.6)

    image = list(result)[0].plot()

    cv2.imshow("test", image)

    key_input = cv2.waitKey(1)

    if key_input == ord('q') &0xff:
        break

cap.release()
cv2.destroyAllWindows()
```
