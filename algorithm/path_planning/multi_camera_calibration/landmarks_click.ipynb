```python
%load_ext autoreload
%autoreload 2
%matplotlib notebook
import matplotlib
matplotlib.use("TkAgg")
import matplotlib.pyplot as plt
import os
import sys
import imageio
import numpy as np
```

```python
def open(image):
    import matplotlib.pyplot as plt
    import numpy as np

    fig = plt.figure(figsize=(10,6))
    ax = fig.add_subplot(111)
    ax.imshow(image)
    text = ax.text(10, 10, "", va="top", ha="left", color='w')

    landmarks = []
    done = {'flag': False}

    def onclick(event):
        # 축 밖 / 좌표 미유효 방지
        if event.inaxes != ax or event.xdata is None or event.ydata is None:
            return
        # 좌클릭만 사용 (노트북에서 우클릭은 종종 안 먹음)
        if event.button == 1:
            landmarks.append([float(event.xdata), float(event.ydata)])
            l = np.float32(landmarks)
            ax.plot(l[:,0], l[:,1], 'r.')
            text.set_text("{}\n{}".format(len(l), np.around(l, 1)))
            fig.canvas.draw_idle()

    def onkey(event):
        # q로 종료
        if event.key == 'q':
            done['flag'] = True
            plt.close(fig)

    def onclose(event):
        done['flag'] = True

    cid1 = fig.canvas.mpl_connect('button_press_event', onclick)
    cid2 = fig.canvas.mpl_connect('key_press_event', onkey)
    cid3 = fig.canvas.mpl_connect('close_event', onclose)

    plt.show()  # notebook backend에선 non-blocking이므로 아래 루프로 대기
    while plt.fignum_exists(fig.number) and not done['flag']:
        plt.pause(0.05)   # 이벤트 루프 돌리기

    # timestamp 대신 extrinsics엔 ids가 필요하므로, 여기선 클릭 순번을 ids로 돌려줄 수도 있음
    ids = list(range(len(landmarks)))
    return landmarks, ids
```
